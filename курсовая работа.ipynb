```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

## подготовка данных

```python
revs = pd.read_csv('/Users/m.makarov/Documents/мирэа тетрал/курсач/IMDB Dataset.csv')

revs_train, revs_test = train_test_split(
    revs,
    test_size=0.2,
    random_state=42,
    stratify=revs['sentiment']
)
revs_train
```

```python
revs['p/n'] = revs['sentiment'].map({'positive': 1, 'negative': 0})     #новый столбец: 1 - позитивый, 2 - негативный
revs
```

```python
revs_train, revs_test = train_test_split(
    revs,
    test_size=0.5,
    random_state=42,
    stratify=revs['sentiment']
)
revs_train
```

## модель наивного байеса

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
model1 = make_pipeline(CountVectorizer(), MultinomialNB())

model1.fit(revs_train['review'], revs_train['p/n'])
```

```python
y_true = revs_test['p/n']
y_true
```

```python
y_pred1 = model1.predict(revs_test['review'])
y_pred1
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score


print('Accuracy: ', accuracy_score(y_true, y_pred1))
print(classification_report(y_true, y_pred1))
print(confusion_matrix(y_true, y_pred1))
```

## logistic regression

```python
from sklearn.linear_model import LogisticRegression

model2 = make_pipeline(TfidfVectorizer(), LogisticRegression())

model2.fit(revs_train['review'], revs_train['p/n'])
```

```python
y_pred2 = model2.predict(revs_test['review'])
y_true = revs_test['p/n']

print('accuracy: ', accuracy_score(y_true, y_pred2))
print(classification_report(y_true, y_pred2))
print(confusion_matrix(y_true, y_pred2))
```

## lirear support vector classification

```python
from sklearn.svm import LinearSVC

model3 = make_pipeline(CountVectorizer(), LinearSVC())
model3.fit(revs_train['review'], revs_train['p/n'])
```

```python
y_pred3 = model3.predict(revs_test['review'])
y_true = revs_test['p/n']

print('accuracy: ', accuracy_score(y_true, y_pred3))
print(classification_report(y_true, y_pred3))
print(confusion_matrix(y_true, y_pred3))
```
